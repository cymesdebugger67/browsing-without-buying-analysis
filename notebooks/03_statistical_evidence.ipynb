# 03 · Statistical evidence and interpretation

**Question:** Which descriptive associations have notable effect sizes, and how uncertain is the new-versus-returning conversion-rate gap? Eight exploratory tests share one Benjamini–Hochberg adjustment family.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

ROOT = Path.cwd()
SOURCE = ROOT / "data" / "raw" / "online_shoppers_intention.csv"
if not SOURCE.is_file():
    raise FileNotFoundError("Run this notebook from the repository root after downloading the UCI CSV")
data = pd.read_csv(SOURCE)
print(f"Loaded {len(data):,} recorded sessions")
```

```python
from scripts.statistical_review import analyze

categorical, numeric, visitor_gap = analyze(data)
display(categorical.sort_values('cramers_v', ascending=False))
```

```python
numeric = numeric.assign(absolute_effect=numeric['rank_biserial_purchasing_vs_nonpurchasing'].abs())
display(numeric.sort_values('absolute_effect', ascending=False))
```

```python
print(f"New visitors: {visitor_gap['new_purchases']:,}/{visitor_gap['new_sessions']:,} ({visitor_gap['new_conversion_pct']:.2f}%)")
print(f"Returning visitors: {visitor_gap['returning_purchases']:,}/{visitor_gap['returning_sessions']:,} ({visitor_gap['returning_conversion_pct']:.2f}%)")
print(f"Difference: {visitor_gap['difference_percentage_points']:+.2f} percentage points")
print(f"Exploratory 95% bootstrap interval: {visitor_gap['bootstrap_95pct_lower_pp']:.2f} to {visitor_gap['bootstrap_95pct_upper_pp']:.2f} pp")
```

**Interpretation:** Tests measure association, not causal effects. Very small adjusted p-values can coexist with modest effect sizes. The session-level bootstrap assumes independent records and cannot independently verify UCI's one-session-per-user design. Inspect chi-square expected-count diagnostics before using asymptotic p-values.
