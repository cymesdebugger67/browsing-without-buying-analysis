# 02 · Where purchases occur

**Question:** How do observed conversion rates vary by recorded visitor type, page depth, and bounce-rate bands? Group size is shown so sparse segments are not mistaken for reliable patterns.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

ROOT = Path.cwd()
SOURCE = ROOT / "data" / "raw" / "online_shoppers_intention.csv"
if not SOURCE.is_file():
    raise FileNotFoundError("Run this notebook from the repository root after downloading the UCI CSV")
data = pd.read_csv(SOURCE)
print(f"Loaded {len(data):,} recorded sessions")
```

```python
from scripts.conversion_deep_dive import prepare_sessions, conversion_table, engagement_table

sessions = prepare_sessions(data)
by_visitor = conversion_table(sessions, ['VisitorType'])
display(by_visitor.sort_values('sessions', ascending=False))
```

```python
by_depth = conversion_table(sessions, ['VisitorType', 'page_depth'])
display(by_depth.sort_values(['VisitorType', 'page_depth']))
```

**Interpretation:** Browsing depth measures observed within-session engagement. Higher purchase rates at larger depths do not show that increasing pages visited would cause a purchase. Segments below 100 sessions are flagged for caution.

```python
by_bounce = conversion_table(sessions, ['bounce_band'])
display(by_bounce)

engagement = engagement_table(sessions)
display(engagement.loc[engagement['metric'].isin(['ProductRelated', 'ExitRates'])])
```

**Limitations:** `PageValues` is omitted because it is transaction-linked. Bounce-rate categories are descriptive bins, not established conversion-risk thresholds. These observations cannot identify cart abandonment or individual shopper intent.
